3.1 Demonstrate through simulation that the sum of many independent normal random variables is also normally distributed, with its mean equal to the sum of the individual means and its standard deviation equal to the square root of the sum of the individual variances. 

Specifically, generate N random values for the means and standard deviations (e.g., between 1 and 50), create 
N independent normal samples using those parameters, and compute their sum. 

Then show that:

* The empirical mean of the summed distribution approximates the theoretical sum of the individual means.
*  The empirical standard deviation approximates the theoretical standard deviation.

```python
# Import necessary packages
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Set random seed
np.random.seed(42)

# Choose how mnay normal random variables we want
N = 10

# Choose how many time to simulate
num_samples = 100000
```

```python
# Randomly Generate N means and Standard Deviations
means = np.random.uniform(1, 50, N)
std_devs = np.random.uniform(1, 50, N)

print("Generated means:")
print(means)
print("\nGenerated standard deviations:")
print(std_devs)
```

```python
# Calculate theoretical mean and standard deviation
theoretical_mean = np.sum(means)
theoretical_std = np.sqrt(np.sum(std_devs**2))

print("\nTheoretical mean of the sum:", theoretical_mean)
print("Theoretical standard deviation of the sum:", theoretical_std)
```

```python
# Generate samples from each of the N normal distributions
samples = np.array([
    np.random.normal(loc=means[i], scale=std_devs[i], size=num_samples)
    for i in range(N)
])

print("\nShape of sample matrix (N rows, num_samples columns):", samples.shape)
```

```python
# Sum the N independent normal variables
sum_samples = np.sum(samples, axis=0)
```

```python
# Calculate empirical mean and standard deviation
empirical_mean = np.mean(sum_samples)
empirical_std = np.std(sum_samples, ddof=1)

print("\nEmpirical mean of the sum:", empirical_mean)
print("Empirical standard deviation of the sum:", empirical_std)
```

```python
# Compare empirical and theoreticl results
mean_error = abs(empirical_mean - theoretical_mean)
std_error = abs(empirical_std - theoretical_std)

print("\nAbsolute error in mean:", mean_error)
print("Absolute error in standard deviation:", std_error)
```

```python
# Plot histogram of summed samples and overlay theoretical normal curve

x = np.linspace(
    theoretical_mean - 4 * theoretical_std,
    theoretical_mean + 4 * theoretical_std,
    1000
)

# Formula for normal density:
# f(x) = (1 / (sigma * sqrt(2*pi))) * exp(-(x - mu)^2 / (2*sigma^2))
theoretical_density = (
    1 / (theoretical_std * np.sqrt(2 * np.pi))
) * np.exp( -((x - theoretical_mean) ** 2) / (2 * theoretical_std ** 2)
)

plt.figure(figsize=(10, 6))
plt.hist(sum_samples, bins=60, density=True, alpha=0.6, label="Simulated summed distribution")
plt.plot(x, theoretical_density, linewidth=2, label="Theoretical normal density")
plt.title("Sum of Independent Normal Random Variables")
plt.xlabel("Summed value")
plt.ylabel("Density")
plt.legend()
plt.show()
```

### The simulation confirms that the sum of independent normal random variables is itself normally distributed. The sample mean of the summed observations closely matches the theoretical mean, which equals the sum of the individual means. Likewise, the sample standard deviation closely matches the theoretical standard deviation, which equals the square root of the sum of the individual variances. The histogram of the simulated sum also aligns well with the theoretical normal density. Beyond this, we can repeat the simulation at multiple levels of N to display that the result will still hold.

3.2 In this exercise, you will explore the Central Limit Theorem (CLT) through simulation. The goal is to observe how the distribution of sample means changes as the sample size increases, even when the underlying population is not normally distributed. You will draw random samples from three different populations:

* a discrete distribution where P(x=1) = .15 and P(x=2) = .85
* an exponential distribution with mean (scale parameter) 0.8
* a triangular distribution with parameters (min, mode, max) = (0, 11, 12).

For each distribution, generate independent samples of size n =5 and n = 40.
Compute the sample mean Z = (X1+....+Xn)/n for each sample, and then create histograms of these sample means. 

Compare the histograms across the two sample sizes to evaluate how the sampling distribution of the mean changes. Discuss whether the shape of the distribution of sample means appears closer to a bell curve (normal distribution) as n increases. 

```python
# Import necessary packages
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Set random seed
np.random.seed(42)

# Choose how many time to simulate
num_repetitions = 10000
```

```python
# Define sample sizes
sample_size = [5,40]
```

```python
# Create functions to simulate sample means
def simulate_discrete_sample_means(n, num_reps):
    """
    Simulate sample means from the discrete distribution:
    P(X=1)=0.15, P(X=2)=0.85
    """
    samples = np.random.choice([1, 2], size=(num_reps, n), p=[0.15, 0.85])
    return samples.mean(axis=1)


def simulate_exponential_sample_means(n, num_reps):
    """
    Simulate sample means from an exponential distribution
    with mean (scale) = 0.8
    """
    samples = np.random.exponential(scale=0.8, size=(num_reps, n))
    return samples.mean(axis=1)

def simulate_triangular_sample_means(n, num_reps):
    """
    Simulate sample means from a triangular distribution
    with parameters (min, mode, max) = (0, 11, 12)
    """
    samples = np.random.triangular(left=0, mode=11, right=12, size=(num_reps, n))
    return samples.mean(axis=1)
```

```python
# Store all simulated sample means
results = {
    "Discrete: P(1)=0.15, P(2)=0.85": {},
    "Exponential: mean=0.8": {},
    "Triangular: min=0, mode=11, max=12": {}
}

for n in sample_size:
    results["Discrete: P(1)=0.15, P(2)=0.85"][n] = simulate_discrete_sample_means(n, num_repetitions)
    results["Exponential: mean=0.8"][n] = simulate_exponential_sample_means(n, num_repetitions)
    results["Triangular: min=0, mode=11, max=12"][n] = simulate_triangular_sample_means(n, num_repetitions)
```

```python
# Plot histograms
fig, axes = plt.subplots(3, 2, figsize=(14, 14))
fig.suptitle("Central Limit Theorem: Sampling Distributions of the Mean", fontsize=16)

distribution_names = list(results.keys())

for row, dist_name in enumerate(distribution_names):
    for col, n in enumerate(sample_size):
        ax = axes[row, col]
        sample_means = results[dist_name][n]
        ax.hist(sample_means, bins=40, density=True, alpha=0.7)

        ax.set_title(f"{dist_name}\nSample size n = {n}")
        ax.set_xlabel("Sample mean")
        ax.set_ylabel("Density")

plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.show()
```

```python
# Print Summary statistics
for dist_name in distribution_names:
    print(f"\n{dist_name}")
    for n in sample_size:
        sample_means = results[dist_name][n]
        print(f"  n = {n}")
        print(f"    Mean of sample means: {np.mean(sample_means):.4f}")
        print(f"    Std. dev. of sample means: {np.std(sample_means, ddof=1):.4f}")
```

### Discrete Distribution: at n = 5, the histogram looks clunky due to the limited number of samples, however, at n = 40, the histogram begins to take more of a bell curve shape suggesting that with more samples the shape will look more normal as n increases.

### Exponential Distribution: at n = 5, the histogram is stronly skewed to the right. At n = 40, the histrogram, again, begins to look more symmetric and bell shaped further demonstrating the CLT.

### Triangular Distribuiton: at n = 5, the histogram is slightly skewed to the left and the histogram takes a more normal curve at n = 40.

### The simulation supports the Central Limit Theorem. For all three underlying populations, the distribution of sample means becomes more bell-shaped as the sample size increases from n = 5 to n = 40. This happens even though the original populations are not normally distributed. In addition, the spread of the sample mean distribution becomes smaller as n increases, reflecting the decrease in the standard error.
